```python
class model(object):
    # constructor method
    def __init__(self,numlayers,numunits,name):
        # the specific instance of this object
        self.layers = numlayers #attributes
        self.units  = numunits
        self.name   = name
        self.weights = 1


    # A function associated w/above class
    def howManyUnits(self):
        totalUnits = self.layers * self.units
        print(f'There are {totalUnits} in the model')
    
    def trainTheModel(self,x):
        print(f"Updating model with {x}")
        self.weights += x
        return self.weights
    def __str__(self):
        return f'This is a {self.name} architecture'
m1 = model(2,5,'CNN')
m1.howManyUnits()
m1.trainTheModel(1.24)
m1.trainTheModel(3)
str(m1)
```

```python
import numpy as np
class another_model(object):
    # constructor method
    def __init__(self,numlayers,numunits,name):
        # the specific instance of this object
        self.layers = numlayers #attributes
        self.units  = numunits
        self.name   = name
        self.weights = np.random.randn(numlayers,self.units)


    # A function associated w/above class
    def howManyUnits(self):
        totalUnits = self.layers * self.units
        print(f'There are {totalUnits} in the model')
    
    def trainTheModel(self,x,y):
        print(f"Updating model with {x}")
        self.weights = self.weights*x + y
        return self.weights
    def __str__(self):
        return f'This is a {self.name} architecture'
m2 = another_model(2,3,'Xnn')
m2.trainTheModel(2,3)
```
