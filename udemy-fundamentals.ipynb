Cross-entropy

$H = - \sum \limits _{i=1}^{n}p(x_{i})log_2(p(x_i))$

x = data values

p = probability

So for a coin flip, n would equal 2. We multiply the probability that an event occurs times the log of that same probability for all possible events. 

*So for coin flip example, **'n'** would be two because we have only two possible outcomes. So we have the probability of heads times the log of the probability of heads PLUS the probability of tails times the log of the probability of tails. We sum those values together, we take the minus sign and entropy is indicated using capital letter 'H'. The minus sign here is because probability values go between zero and one, and the log of a number between zero and one is negative. So we have to flip it to positive by using the negative sign*

<font color="orange">High entropy meanas that the dataset has a lot of variability.</font>

<font color="yellow">Low entropy means that most of the values of the dataset repeat (and therefore are redundant)</font>

<font color="orange">How does entroyp differ from variance?</font>

<font color="yellow">Note entropy and variance are two different mesaures of variability. Entropy is nonlinear and makes no assumptions about the distribution.</font>

<font color="teal">Variance depends on the validity of the mean and therefore is appropriate for roughly normal data.</font>

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p = .25 # probability of event happening
H = -( p*np.log(p))
# The reason its wrong, we ONLY calculating prob of 1 event.
# There is also the prob of event NOT happening
print('Wrong entropy: ' + str(H))
```

```python
# The correct way
x = [.25,.75]
H = 0
for p in x:
    H += -p*np.log(p)
print("Correct entropy: " + str(H))
```

```python
# Also corrrect, this is also known as 'Binary cross-entropy'
# One of the most important loss functions in DL (deep learning)
H = -( p *np.log(p) + (1-p)*np.log(1-p)) # the formula for exactly 2 events
print('correct entroy: ' + str(H))
```

Cross-Entropy -- <font color="orange">The formula is basically the same</font>

```python
# note: all probs must sum to 1!
p = [1,0] # sum=1  seems like the true probabilites IMU
q = [.25,.75] #sum =1 is what the model is.
# IMO it seems we want 'q' to exactly like 'p' in a great world

H = 0
for i in range(len(p)):
    H -=p[i]*np.log(q[i])

print ('Cross entropy: ' + str(H))
```

```python
# also correct, written out for N=2 events
H = -( p[0]*np.log(q[0]) + p[1] * np.log(q[1]))
print('Cross Entropy again: ' + str(H))
# In binary cross-entropy, when we know that one of the probs is 0, we can simplify to
H = -np.log(q[0])  # cause note q[1] is 0 and thus the whole multiplicatin of that term is 0
print("Manually simplified: " + str(H))
```
