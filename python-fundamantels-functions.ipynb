```python
import numpy as np
import pandas as pd 

# create random data
# returns a random number from a Gaussian distribution i.e. normal distribution. that is what the 'n' in randn is for
np.random.randn() 
# normal dist has a mean of 0 and standard deviation of 1
np.random.randn(5,5)
np.random.randn(5,5)*5 + 20
var1 = np.random.randn(100)*5 + 20
var2 = np.random.randn(100) > 0

# variable labels
labels = ['Temp (C)','Ice Cream']
# create a dictionary to put them together
D = {labels[0]:var1,labels[1]:var2}
#D
# Import the dictionary into a pandas dataframe

df = pd.DataFrame(data=D)
type(df)
df
```

```python
 # note ice cream is boolean but pandas knows to convert them to 0 and 1 and the mean is around .5
# meaning 1/2 the labels are false and half are true
df.mean() 
```

```python
# Ex: Create  a fcn that computes a factorial
import math

def myFactorial(n):
    return np.prod(np.arange(1,n+1))

y = 3
myFactorial(y),math.factorial(y)
```
