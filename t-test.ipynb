And when running experiments, one of the things that we need to do is determine whether one particular model architecture or a set of meta-parameters is better than a different architecture or a different set of meta-parameters. 

So how do we determine whether our experiment was successful?
How do we determine whether one model architecture is significantly better than another?

So the answer to that question is a t-test. What is a t-test? A t-test is way to determine whether data drawn from one distribution is significantly different from data drawn from another distribution or sample. 

$H_{0}$ Stands for the null hypothesis. The null hypothesis states that the two models perform equally well. This means we can change accuracies between the two models and that wouldn't change the average accuracy. 

The goal of the t-test is to provide evidence AGAINST the null hypothesis. If there's strong enough evidence against the null hypothesis, then we go with the $H_{A}$ hyothesis iow: the alternative hypothesis.
