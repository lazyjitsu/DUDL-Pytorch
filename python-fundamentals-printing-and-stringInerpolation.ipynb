```python
## String Interpolation
num = 11.7
sng = 'cellar doors'
# alittle long and tedious
print('I would like to have ' + str(num) + ' ' + sng + '.')
# more compact/readable
print('I would like to have %g %s ' %(num,sng))
print('I would like to have %f %s ' %(num,sng))
print('I would like to have %.3f %s ' %(num,sng))
print('I would like to have %15.3f %s ' %(num,sng)) # 15 spaces before float#

```

```python
# f-strings
print(f'I would like to have {num:.2f} {sng}')
# similar
print(f'another way {num:.2f} {sng}'.format(num=num,sng=sng))
#same
print(f'another way {num:.2f} {sng}'.format(n=num,s=sng))
```

```python
import matplotlib.pyplot as plt 
import numpy as np

plt.subplot(1,2,1)
plt.plot(np.random.randn(5))

plt.subplot(1,2,2)
plt.plot(np.random.randn(5))

plt.show()
```

```python
# subplots is better than subplot

fig,ax = plt.subplots(1,2)
x = np.arange(10)
ax[0].plot(x,x**2,'b')
ax[0].plot(x,x,'r')

ax[1].plot(x,np.sqrt(x),'r')
plt.show()
```

```python
# we can use figsize w/tuples (width,height)

fig,ax = plt.subplots(2,2,figsize=(15,3))
# ax.shape
ax[0,0].plot(np.random.randn(4,4))
ax[0,1].plot(np.random.randn(4,4))
ax[1,0].plot(np.random.randn(4,4))
ax[1,1].plot(np.random.randn(4,4))
plt.tight_layout()
plt.show()

```

```python
# Making graphs look nicer
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3,3,101)
plt.plot(x,x,label='y=x')
plt.plot(x,x**2,label='y=x**2')
plt.plot(x,x**3,label='y=x**3')
plt.legend()
plt.xlabel("X",fontsize=12)
plt.ylabel("Y=f(x)")

plt.title("A coool Plot")
# plt.xlim([-3,3]) which is hardcoded for -3 and 3 lets try soft coding
plt.xlim(x[0],x[-1])
plt.ylim(-10,10)
# plt.axis('square')
# Better to use gca() - get current axis--because square may not work/look correct--sets the aspect ratio--
# IOW: the size/distance of the ticks on x and y axis
# use 1 if you want the distance of the ticks to be equal on the x and y axis. 'auto' is the same as not having it set
plt.gca().set_aspect(.3) 
plt.show()

```

```python
# Subplots ; Better than subplot
fig,ax = plt.subplots(2)

x = np.linspace(-3,3,101)
ax[0].plot(x,x,label='y=x')
ax[0].plot(x,x**2,label='y=x**2')
ax[0].plot(x,x**3,label='y=x**3')
ax[0].legend()
ax[0].set_xlabel("X",fontsize=12)
ax[0].set_ylabel("Y=f(x)")
ax[0].set_xlim([x[0],x[-1]])
ax[0].set_ylim([-10,10])

ax[0].set_title("A cool Plot 0")

ax[1].plot(x,x,label='y=x')
ax[1].plot(x,x**2,label='y=x**2')
ax[1].plot(x,x**3,label='y=x**3')
ax[1].legend()
ax[1].set_xlabel("X",fontsize=12)
ax[1].set_ylabel("Y=f(x)")
ax[1].set_title("A cool Plot 1")
```

```python
fig,ax = plt.subplots(2,3,figsize=(15,7))

x = np.linspace(-3,3,101)
ax[0,0].plot(x,x,label='y=x')
ax[0,0].plot(x,x**2,label='y=x**2')
ax[0,0].plot(x,x**3,label='y=x**3')
ax[0,0].legend()
ax[0,0].set_xlabel("X",fontsize=12)
ax[0,0].set_ylabel("Y=f(x)")
ax[0,0].set_xlim([x[0],x[-1]])
ax[0,0].set_ylim([-10,10])

ax[0,1].plot(x,x,label='y=x')
ax[0,1].plot(x,x**2,label='y=x**2')
ax[0,1].plot(x,x**3,label='y=x**3')
ax[0,1].legend()
ax[0,1].set_xlabel("X",fontsize=12)
ax[0,1].set_ylabel("Y=f(x)")
ax[0,1].set_xlim([x[0],x[-1]])
ax[0,1].set_ylim([-10,10])

# make sure the subplots are not laying on top of ea. other
plt.tight_layout() 
plt.show()
```

```python
# Ex: make a cool charrt with ascending lines. 
import numpy as np
fig,ax = plt.subplots()
x = np.linspace(0,10,100)

for i in np.linspace(0,1,50):
    # print(i)
    #              color=r,g,b 
    ax.plot(x,x**i,color=[i/2,0,i])
plt.show()
```

```python
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd 

import seaborn as sns

n = 200

D = np.zeros((n,2))
# the first col will be linearly spaced numbers from 0 to 10 in N steps (i.e 200 steps) + some random noise
D[:,0] = np.linspace(0,10,n) + np.random.randn(n)

D[:,1] = D[:,0]**2 + np.random.randn(n)*10
 #  a hack cause sns don't work same as Udemy's version
sns.jointplot(x=D[:,0],y=D[:,1])
# the way plt would plot it
#plt.plot(D[:0,],D[:,1]) 
plt.show()
```

```python
df = pd.DataFrame(data=D,columns=['var1','var2'])
df
# sns.jointplot(x=df.columns[0],y=df.columns[1],data=df)
# recall a 'kde' stands for kernel densitiy estimator and is used to estimate the probability density
sns.jointplot(x=df.columns[0],y=df.columns[1],data=df,kind='kde')

plt.show()
```
