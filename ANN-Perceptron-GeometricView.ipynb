A linear operating just means that you're doing addition and scalar multiplication. Non-linear is anything else.

**Linear models** only solve linearly seperable problems. Linear seperable probems are problems where you can achieve the result by drawing a stragiht line through the two different groups. Linear models are not good at solving non-linear problems.

**Non-linear Models** can solve more complex problems.

<font color="yellow">Never use a linear model for non linear problem and never use a non-linear model to solve a linear problem!</font>

Now, linear models are what you come across in a standards course on statistics and machine learning, things like regression, general linear models, ANOVA, these are all, exempt factor analysis. These are all examples of linear models.

Now, linear models are really great and there are lots and lots of linearly separable problems in the universe, and we use linear models for them, any kind of general variant of a general linear model.

But there are also many problems that cannot be solved through linear models, and that's why we need non-linear operations
in our deep learning models. And then also the second part of the statement, not to use a non-linear model for a linear problem.
