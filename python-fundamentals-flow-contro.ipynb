```python
# Exercise: Poisson counter. Note a poisson counter is used to simulate the random firing of brain cells by counting
# the number of times it takes to go from 1 to some target value
# to create a poison counter
# 1) Specifiy a target: exp(-lambda) i.e. e^-lambda and we'll make lambda = 10
# 2) initalize a counter = 0, currval = 1
# 3) Multiply currval by a randon number U[0,1] i.e a Uniform Distribution
# 4) Repeat Step 3 until currval < target
# 5) return # fo times that you go through step 4
import numpy as np

def poissonCounter(lam):
    counter,currval = 0,1
    target = np.exp(-lam)

    while currval > target:
        counter += 1
        # rand() is the uniform dist
        currval *= np.random.rand()

    return counter

poissonCounter(10)
```
