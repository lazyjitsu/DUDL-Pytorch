<font color="teal" size =10> Indexing and Slicing

```python
anotherList = [2,3,4,'qwerty',33]

idx = 4.0
# anotherList[idx] #will not work
anotherList[int(idx)]  # however, will work
```

```python
anotherList[len(anotherList)-1] # get the last item in the list
list1 = [4,'hi',[5,4,3],'yo',{'Squirrel':'cute','Penguin':'Yummy'}]
# Get Penguin attribute
list1[4]['Penguin']
```

```python
a = list(range(5,11)) # convert the range 5,11 to a list
a
```

```python
a[0:3] # non-inclusive
a[:3] # we can drop the zero as it's implied
```

```python
#  a[start:stop:skip]
# lets say we wanted every other element, we can use '::'
a[::2] # all elements skipping by 2
a[1::2] # still skipping by 2 but start at first el

```

```python
# reverse a list
a[::-1]
```
