<font color="teal" size = 8> Vanishing and Exploding gradients</font>

We already know about overshooting. Remember, we alwasy move in a direction that is opposite to the derivative at that point AND the step size, the amounts that we move is proprotional to the magnitude of the derivative. Eventually the steps get smaller and smaller until we simply stop learning. This is known as 'Vanishing Gradient'


![alt text](img/vanishing-gradient.png) 

**Vanishing Gradient**  Weights don't change --> no learning and is problematic for DL

**Exploding Gradient**   Weights change wily -> bad solutions

<font color="orange" size=7> How to minimize gradient problems</font>

- Use models w/few hidden layers
- Use activation functions that do not saturate (e.g. ReLU)
- Apply weight normalization
- Pre-train networks using autoencoders
- Use regularization techniques like batch normalization, dropout and weight decay
- Use architectures like **residual networks** ("resnet")
