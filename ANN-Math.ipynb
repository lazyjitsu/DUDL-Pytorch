## Math Part 3 (Backprop)

Although we call the new notation a 'unit' others may call it a 'neuron'
![alt text](img/math-3-1.png)

Now you can see the output of the previous layer is the input to the next. Though these nodes give multipleoutputs, in fact, its the exact same output, the same numerical value, its just copied to multiple units. Each of these nodes is doing the same calculation:

![alt text](img/math-3-2.png)

Each node believes its alone in the universe. They all think they're alone. 

This is important to know because we need to adjust the weights for each node indpendently. 

**Forward propagation** Comppute output based on input

**Backwards propagation (backprop)** Adjust the weights based on loss/cost

So the error at the end of the network, at $\hat{y}$  is where it is computed and is used for learning and because it's used for learning it needs to propagate backwards through the network (to the left of $\hat{y}$) to reach all the nodes. 

Backpropagation is just gradient descent. There is a little more involved because we have a high dimensional feature space and we have embedded functions so we're gonna need touse the chain rule of derivatives.E.g.
$$F^{\prime}(x) = f^{\prime}(g(x))*g^{\prime}(x)$$
But just note the concept is the same. 
![alt text](img/math-3-3.png)

The idea is to keep moving along this x axis until we find the set of weights that minimized our loss or our cost function. I believe the x axis is now going to contain the weights IMU. We just move along this axis, the x-axis, aka the 'weight axis' until we find the set of wights that minimizes that minimizes our loss or our cost function. 

So lets to through backprop step by step:

The math taking place in ea. node. The X is fixed and the only thing we can change are the weights, W: 

$$ \sigma(x^{T}w) \rightarrow\hat{y} $$

$$w \leftarrow w - \eta\partial L$$

$\eta$ is Learing rate pronounced "eta"

$\partial$ is the partial deriviative (i.e. for one dimension)

L is Deriviative of loss

So this little partial L over here actually expands to DL by DW or partial L, partial W, which we can expand further to replace y hat with the computation that is actually taking place inside each node.

&emsp;&emsp;$\frac{\partial L(\hat{y},y)}{\partial w}$ = $\frac{\partial L(\sigma(x^Tw),y)}{\partial w}$

How do we calculate this deriviate: $\frac{\partial L(\sigma(x^Tw),y)}{\partial w}$

Recall the loss function for MSE: $L = \frac{1}{2}(\hat{y}-y)^2$

Now, exactly how we proceed from this point depends on the activation function, because each activation function has its own derivative. And that working through all of the details of the math gets pretty complicated pretty quickly, because it depends on a variety of choices that you make when constructing the model, including which loss function you use and which activation function is computed inside each node. 

![alt text](img/back-prop-1.png)

And I would like us to focus more on the principle here, that this expression, updating the weights according to the derivative of the loss function with respect to the weights times the learning rate, or eta, is the same thing as gradient descent. Also, note that I'm showing everything here with the loss function, but of course, you can simply substitute the cost function for the loss function, which doesn't actually change the idea.

It just means that we're averaging over a collection of samples instead of running back propagation one sample at a time. Again, it's going to make everything one layer more complicated. It adds a little bit more complexity.

It doesn't change the fundamental principle here. But even if I went through all the calculus, that wouldn't still fully explain
how it actually works in practice, because the actual code implementations are a little bit different from the formulas that I showed, due to the potential for numerical instabilities and also from computation time issues that arise from implementing the calculus directly. So in fact, the developers of PyTorch and other deep learning toolboxes have come up with some computational tricks to make the back propagation procedure fast and stable.

But again, the concept is exactly the same as what you learned about in gradient descent.
