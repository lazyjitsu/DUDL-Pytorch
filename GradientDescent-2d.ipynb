So we know if we have a function that looks like the top left pic, we can compute the derivative which is hte slop of the function at any given point.

Now when we move to two dimensional functions (above right fcn pic), there is a derivative along each dimension or each axis. So there is a derivative on dimension 'x' another on dimension 'y' and so on. At any given point we can compute what's called the partial derivatives.  The partial derivatives are the derivatives of the function, ignorning one dimension and only focusing on th eother dimension. So the partial derivative with respect to X, means that we ignore Y, and we just see how is the function changing over X. And similar for Y and so we have two dimensions means we have two partial derivatives. The case on the left we can call it a partial derivative w/respect to x but there is no other variable to be partial to so we just call it a derivative. Ok, so if we have two partial derivatives (taking the top right example for the moment) what we can do is put both derivatives in a list, a two element vector, an array of partial derivatives, and that's what we call the gradient!
![alt text](DUDL-Pytorch/img/gx-2.GIF)

So the gradient of the function at any given point is just a collection of all the partial derivatives w/respect toa ll the dimensions of that function. So the gradient is basically the same as the derivative except it's involving more numbers because we have a higher dimensionality. 

Note all the different notations you may see. The 'a' looking one is called a **del**. The upside down triangle is called a **nabla** and is just a collection of all the partial derivatives with respect to all of the different directions or dimensions of the data.
